```python
#Missing data handling: choose an appropriate strategy for each column
#(mean/median imputation, mode imputation, forward fill, or row deletion) and justify
#each choice in a markdown cell
# 1. Droping rows where Price is missing
df = df.dropna(subset=['Price'])
#Price is the target variable for property evaluation.
#Imputing synthetic target values creates artificial bias for analysis and modeling.
```

```python
# 2. Fill counts with Mode (Most Frequent)
df['Car'] = df['Car'].fillna(df['Car'].mode()[0])
df['Bathroom'] = df['Bathroom'].fillna(df['Bathroom'].mode()[0])
df['Bedroom2'] = df['Bedroom2'].fillna(df['Bedroom2'].mode()[0])
#Discrete counts of property features. Using the mode ensures values remain valid whole numbers.
```

```python
# 3. Fill missing text with Mode
df['CouncilArea'] = df['CouncilArea'].fillna(df['CouncilArea'].mode()[0])
df['Regionname'] = df['Regionname'].fillna(df['Regionname'].mode()[0])
#Categorical location features. Filled using the most frequent regional category.
```

```python
# 4. Fill numbers with Median
df['Landsize'] = df['Landsize'].fillna(df['Landsize'].median())
df['BuildingArea'] = df['BuildingArea'].fillna(df['BuildingArea'].median())
df['Distance'] = df['Distance'].fillna(df['Distance'].median())
df['YearBuilt'] = df['YearBuilt'].fillna(df['YearBuilt'].median())
#Continuous physical dimensions and distances are right-skewed with extreme outliers. Median prevents distortion.
```

```python
# 5. Fill location coordinates with Median
df['Lattitude'] = df['Lattitude'].fillna(df['Lattitude'].median())
df['Longtitude'] = df['Longtitude'].fillna(df['Longtitude'].median())
df['Postcode'] = df['Postcode'].fillna(df['Postcode'].mode()[0])
#Spatial coordinates are best imputed using the median coordinates of the corresponding Suburb.
```

```python
# 6. Forward fill a single column
df['Car'] = df['Car'].ffill()
```

```python
#Duplicate removal: identify and remove duplicate rows; document how many were removed
print(df.duplicated().sum())
```

```python
# Define the columns
num_cols = ['Price', 'Landsize', 'BuildingArea', 'Distance']

# Plot boxplots for all columns
plt.figure(figsize=(12, 6))
sns.boxplot(data=df[num_cols])
plt.xticks(rotation=45)
plt.title("Outlier Check Across Multiple Columns")
plt.show()
```

```python
# Function to cap outliers using IQR method
def cap_outliers(df, column_name):
    Q1 = df[column_name].quantile(0.25)
    Q3 = df[column_name].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Cap values outside the boundaries
    df[column_name] = df[column_name].clip(lower=lower_bound, upper=upper_bound)
    return df

# Apply capping to 'Price'
df = cap_outliers(df, 'Price')
df
```

```python
#Data type correction: ensure all columns have the correct dtype (e.g., dates as datetime, IDs as string, monetary values as float)
# 1. Convert Date column to datetime format
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# 2. Convert IDs and Postcodes to String/Object (not numeric)
df['Postcode'] = df['Postcode'].astype(str)

# 3. Convert Monetary and Continuous values to Float
float_cols = ['Price', 'Landsize', 'BuildingArea', 'Distance', 'Lattitude', 'Longtitude']
df[float_cols] = df[float_cols].astype(float)

# 4. Convert Discrete Counts to Integer
int_cols = ['Rooms', 'Bedroom2', 'Bathroom', 'Car', 'Propertycount']
df[int_cols] = df[int_cols].astype(int)
```
